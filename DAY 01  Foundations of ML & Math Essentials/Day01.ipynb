```python
import numpy as np
import pandas as pd

# Ensure the columns needed exist and handle potential data type issues
# 'average_rain_fall_mm_per_year' can sometimes be parsed as object if there are strings or non-numeric formatting
df5['average_rain_fall_mm_per_year'] = pd.to_numeric(df5['average_rain_fall_mm_per_year'], errors='coerce')
df_clean = df5[['hg/ha_yield', 'average_rain_fall_mm_per_year']].dropna()

# Convert to a NumPy array
data = df_clean.to_numpy()

# Set random seed for reproducibility
np.random.seed(42)

# Shuffle indices to perform a random split
indices = np.arange(len(data))
np.random.shuffle(indices)

# Calculate split index for 80/20
split_idx = int(0.8 * len(data))

train_idx, test_idx = indices[:split_idx], indices[split_idx:]
train_data, test_data = data[train_idx], data[test_idx]

# Separate back into yield and rainfall
yield_train, rain_train = train_data[:, 0], train_data[:, 1]
yield_test, rain_test = test_data[:, 0], test_data[:, 1]

print(f"Total samples: {len(data)}")
print(f"Train samples: {len(train_data)}")
print(f"Test samples: {len(test_data)}")
```

```python
# Calculate stats for Yield and Rainfall on the training set using pure NumPy
yield_mean = np.mean(yield_train)
yield_var = np.var(yield_train)
yield_std = np.std(yield_train)

rain_mean = np.mean(rain_train)
rain_var = np.var(rain_train)
rain_std = np.std(rain_train)

print("=== Crop Yield (hg/ha_yield) Training Stats ===")
print(f"Mean:      {yield_mean:.2f}")
print(f"Variance:  {yield_var:.2f}")
print(f"Std Dev:   {yield_std:.2f}\n")

print("=== Rainfall (average_rain_fall_mm_per_year) Training Stats ===")
print(f"Mean:      {rain_mean:.2f}")
print(f"Variance:  {rain_var:.2f}")
print(f"Std Dev:   {rain_std:.2f}")
```
